```python
# =====================================================================
# TAHAP 2.1 — PENGUMPULAN DATA
# =====================================================================
import os, glob, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import silhouette_score, davies_bouldin_score
try:
    from IPython.display import display
except ImportError:
    display = print

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 200)
pd.set_option("display.float_format", lambda v: f"{v:,.4f}")
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams["figure.dpi"] = 100

# ---------------------------------------------------------------------
# [1.0] KONFIGURASI EKSPERIMEN (ubah di sini bila diperlukan)
# ---------------------------------------------------------------------
DATA_PATH    = "data/Filedata_Data_Meteorologi.csv"   # letakkan CSV di folder yang sama dengan notebook
FEATURES     = ["grad", "temp_air", "hum_air", "ff", "rain"]
RANDOM_STATE = 42
SIL_SAMPLE   = 10000      # ukuran sampel Silhouette per stasiun (None = hitung eksak, sangat lambat pada S0)
# Parameter DBSCAN = VARIABEL KONTROL (default scikit-learn), dikunci dan sama untuk S0, S1, S2, S3
EPS          = 0.5        # ε (radius tetangga)
MIN_PTS      = 5          # MinPts (jumlah minimum titik)
OUT_DIR      = "hasil_eksperimen"
os.makedirs(OUT_DIR, exist_ok=True)

if not os.path.exists(DATA_PATH):                    # cari otomatis bila nama file berbeda
    kandidat = glob.glob("*Filedata Data Meteorologi*.csv") + glob.glob("/content/*Data_Meteorologi*.csv")
    if not kandidat:
        raise FileNotFoundError("File CSV tidak ditemukan. Letakkan di folder notebook atau ubah DATA_PATH.")
    DATA_PATH = kandidat[0]

# ---------------------------------------------------------------------
# [1.1] MEMUAT DATASET (CSV dari data.go.id; encoding utf-8-sig menghapus BOM)
# ---------------------------------------------------------------------
df = pd.read_csv(DATA_PATH, encoding="utf-8-sig")
print(f"Sumber data : {DATA_PATH}")
print(f"Jumlah baris: {df.shape[0]:,} | Jumlah kolom: {df.shape[1]}")

# [1.2] CONTOH 5 DATA TERATAS (Tabel 1 pada proposal)
print("\n[Tabel 1] Lima data teratas")
display(df.head(5))

# [1.3] DESKRIPSI VARIABEL (Tabel 2 pada proposal)
deskripsi = pd.DataFrame({
    "Nama variabel": ["periode_data", "tanggal", "stasiun_pengamatan", "grad", "temp_air", "hum_air", "ff", "rain"],
    "Keterangan": ["Periode pengamatan (YYYYMM)", "Tanggal pengamatan", "Identitas stasiun meteorologi",
                   "Radiasi", "Temperatur udara", "Kelembapan udara", "Kecepatan angin", "Curah hujan"],
    "Tipe data": [str(df[c].dtype) for c in df.columns],
    "Peran": ["Informasi waktu", "Informasi waktu", "Pemisah lokasi",
              "Fitur clustering", "Fitur clustering", "Fitur clustering", "Fitur clustering", "Fitur clustering"],
})
print("\n[Tabel 2] Deskripsi variabel")
display(deskripsi)

print("\nPeriode data :", df["periode_data"].min(), "s.d.", df["periode_data"].max())
print("Jumlah varian penulisan nama stasiun:", df["stasiun_pengamatan"].nunique())

# =====================================================================
# TAHAP 2.2 — AUDIT DATA DAN EKSPLORASI DATA (EDA)
# =====================================================================
# ############################  A. AUDIT DATA  ############################

# ---------------------------------------------------------------------
# [A1] STRUKTUR DATASET
# ---------------------------------------------------------------------
print("=" * 70, "\n[A1] STRUKTUR DATASET\n" + "=" * 70)
df.info()

# ---------------------------------------------------------------------
# [A2] KONSISTENSI IDENTITAS STASIUN
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[A2] KONSISTENSI IDENTITAS STASIUN\n" + "=" * 70)
varian = df["stasiun_pengamatan"].map(repr).value_counts()      # repr agar karakter tersembunyi (mis. \t) terlihat
print(f"Jumlah varian penulisan nama stasiun: {len(varian)}")
display(varian.rename("jumlah_baris").to_frame())
audit_kode = df["stasiun_pengamatan"].str.extract(r"(DKI\d)")[0]
print("Baris tanpa kode stasiun terdeteksi:", audit_kode.isna().sum())
print("Jumlah lokasi unik berdasarkan kode :", audit_kode.nunique(), "->", sorted(audit_kode.unique()))

# ---------------------------------------------------------------------
# [A3] VALIDITAS TANGGAL (periode_data = YYYYMM, tanggal = hari)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[A3] VALIDITAS TANGGAL\n" + "=" * 70)
def bentuk_tanggal(d):
    return pd.to_datetime(dict(year=d["periode_data"] // 100, month=d["periode_data"] % 100, day=d["tanggal"]),
                          errors="coerce")
audit_tgl = bentuk_tanggal(df)
tgl_invalid = df[audit_tgl.isna()]
print(f"Jumlah baris dengan tanggal tidak valid: {len(tgl_invalid)}")
display(tgl_invalid)

# ---------------------------------------------------------------------
# [A4] DATA DUPLIKAT
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[A4] DATA DUPLIKAT\n" + "=" * 70)
kolom_dup = ["periode_data", "tanggal", "stasiun_pengamatan"] + FEATURES
print("Duplikat identik (periode+tanggal+stasiun+5 variabel):", df.duplicated(subset=kolom_dup).sum())
print("Duplikat seluruh kolom                              :", df.duplicated().sum())

# ---------------------------------------------------------------------
# [A5] NILAI HILANG (MISSING VALUE)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[A5] NILAI HILANG\n" + "=" * 70)
mv = df[FEATURES].isna().sum().rename("jumlah_missing").to_frame()
mv["persen"] = mv["jumlah_missing"] / len(df) * 100
display(mv)
baris_mv = df[FEATURES].isna().any(axis=1).sum()
print(f"Baris dengan >=1 nilai kosong : {baris_mv:,} ({baris_mv / len(df) * 100:.2f}%)")
print(f"Baris lengkap pada 5 variabel : {len(df) - baris_mv:,}")
print("Variabel dengan missing terbanyak:", mv["jumlah_missing"].idxmax())

# ---------------------------------------------------------------------
# [A6] NILAI TIDAK WAJAR
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[A6] NILAI TIDAK WAJAR\n" + "=" * 70)
m_temp = df["temp_air"] < -30
m_ff   = df["ff"] > 100
m_rain = df["rain"] > 300
print(f"temp_air < -30 °C (sentinel ≈ -39.999) : {m_temp.sum()} baris | stasiun: {audit_kode[m_temp].value_counts().to_dict()}")
print(f"ff > 100 m/s                           : {m_ff.sum()} baris | periode: {sorted(df.loc[m_ff, 'periode_data'].unique())}")
print(f"   ringkasan ff > 100: min={df.loc[m_ff, 'ff'].min():.1f}, median={df.loc[m_ff, 'ff'].median():.1f}, max={df.loc[m_ff, 'ff'].max():.1f}")
print(f"   ff pada rentang 20-100 m/s          : {((df['ff'] > 20) & (df['ff'] <= 100)).sum()} baris (dipertahankan, diperiksa di EDA)")
print(f"rain > 300 mm                          : {m_rain.sum()} baris")
display(df[m_rain])
print("   -> Catatan: nilai rain = 350.06 identik dengan nilai ff pada baris yang sama (indikasi salah catat/kolom tertukar).")
print("   rain terbesar lainnya (kuantil 99.9%):", df.loc[~m_rain, "rain"].quantile(0.999).round(2), "mm")
print("Nilai negatif: grad =", (df["grad"] < 0).sum(), "| ff =", (df["ff"] < 0).sum(), "| rain =", (df["rain"] < 0).sum())
print("hum_air di luar 0-100 %:", ((df["hum_air"] < 0) | (df["hum_air"] > 100)).sum())

# ############################  B. EKSPLORASI DATA (EDA)  ############################
# Salinan khusus EDA: nilai tidak wajar pada A6 disembunyikan (NaN) agar grafik terbaca.
# Data asli `df` TIDAK diubah; penanganan resmi dilakukan pada tahap Pra-Pemrosesan.
eda = df.copy()
eda["kode"] = audit_kode
eda["tgl"]  = audit_tgl
eda.loc[m_temp, "temp_air"] = np.nan
eda.loc[m_ff,   "ff"]       = np.nan
eda.loc[m_rain, "rain"]     = np.nan
eda = eda.dropna(subset=["tgl"])
warna_stasiun = dict(zip(sorted(eda["kode"].unique()), sns.color_palette("colorblind", 5)))

# ---------------------------------------------------------------------
# [B1] STATISTIK DESKRIPTIF
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[B1] STATISTIK DESKRIPTIF (setelah nilai tidak wajar disembunyikan)\n" + "=" * 70)
display(eda[FEATURES].describe().T)
print("Statistik deskriptif per stasiun (rata-rata):")
display(eda.groupby("kode")[FEATURES].mean())

# ---------------------------------------------------------------------
# [B2] DISTRIBUSI NILAI & KEMENCENGAN (SKEWNESS)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[B2] DISTRIBUSI NILAI\n" + "=" * 70)
fig, axes = plt.subplots(2, 3, figsize=(15, 7))
for ax, f in zip(axes.ravel(), FEATURES):
    ax.hist(eda[f].dropna(), bins=60, color="#4C72B0")
    ax.set_yscale("log"); ax.set_title(f"Distribusi {f} (skala log pada sumbu y)")
    ax.set_xlabel(f); ax.set_ylabel("frekuensi")
axes.ravel()[-1].axis("off")
plt.tight_layout(); plt.show()
print("Skewness tiap variabel:")
display(eda[FEATURES].skew().rename("skewness").to_frame())

# ---------------------------------------------------------------------
# [B3] PERUBAHAN NILAI BERDASARKAN TANGGAL (rata-rata harian; rain = total harian)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[B3] PERUBAHAN NILAI BERDASARKAN TANGGAL\n" + "=" * 70)
agg_h = {f: ("sum" if f == "rain" else "mean") for f in FEATURES}
harian = eda.groupby(["kode", "tgl"]).agg(agg_h).reset_index()
fig, axes = plt.subplots(5, 1, figsize=(15, 15), sharex=True)
for ax, f in zip(axes, FEATURES):
    for k, g in harian.groupby("kode"):
        ax.plot(g["tgl"], g[f], lw=0.7, label=k, color=warna_stasiun[k])
    ax.set_ylabel(f + (" (total/hari)" if f == "rain" else " (rata-rata/hari)"))
axes[0].legend(ncol=5, loc="upper right"); axes[0].set_title("Deret waktu harian per stasiun")
plt.tight_layout(); plt.show()

# ---------------------------------------------------------------------
# [B4] PERBANDINGAN KARAKTERISTIK ANTARSTASIUN
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[B4] PERBANDINGAN ANTARSTASIUN\n" + "=" * 70)
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, f in zip(axes.ravel(), FEATURES):
    sns.boxplot(data=eda, x="kode", y=f, order=sorted(warna_stasiun), palette=warna_stasiun, showfliers=False, ax=ax)
    ax.set_title(f"{f} per stasiun (tanpa pencilan)"); ax.set_xlabel("")
axes.ravel()[-1].axis("off")
plt.tight_layout(); plt.show()

# ---------------------------------------------------------------------
# [B5] KORELASI ANTARVARIABEL (hubungan linear, bukan sebab-akibat)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[B5] CORRELATION HEATMAP\n" + "=" * 70)
plt.figure(figsize=(6.5, 5))
sns.heatmap(eda[FEATURES].corr(), annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1, square=True)
plt.title("Korelasi Pearson antarvariabel meteorologi"); plt.tight_layout(); plt.show()

# ---------------------------------------------------------------------
# [B6] JUMLAH PENGAMATAN PER TANGGAL DAN STASIUN (pola kelengkapan data)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[B6] POLA KELENGKAPAN DATA\n" + "=" * 70)
n_obs = eda.groupby(["kode", "tgl"]).size().rename("n_obs").reset_index()
print("Distribusi jumlah baris per stasiun per hari (10 terbanyak):")
display(n_obs["n_obs"].value_counts().head(10).rename("jumlah_hari").to_frame())
modus_obs = int(n_obs["n_obs"].mode()[0])
print(f"Jumlah pengamatan paling sering per stasiun per hari = {modus_obs}  -> acuan kelengkapan periode (48 baris/hari)")
print("Jumlah hari berbeda per stasiun:", n_obs.groupby("kode")["tgl"].nunique().to_dict())
fig, axes = plt.subplots(1, 2, figsize=(15, 4))
for k, g in n_obs.groupby("kode"):
    axes[0].plot(g["tgl"], g["n_obs"], lw=0.7, label=k, color=warna_stasiun[k])
axes[0].axhline(48, color="k", ls="--", lw=0.8); axes[0].set_title("Jumlah pengamatan per hari"); axes[0].legend(ncol=5)
axes[1].hist(n_obs["n_obs"], bins=range(0, 60), color="#4C72B0")
axes[1].set_yscale("log"); axes[1].set_title("Distribusi jumlah pengamatan per hari"); axes[1].set_xlabel("n pengamatan")
plt.tight_layout(); plt.show()

print("\n>> KESIMPULAN AUDIT: 20 varian nama stasiun, 1 tanggal tidak valid, tidak ada duplikat, missing terbanyak pada hum_air,"
      "\n   serta nilai tidak wajar pada temp_air, ff, dan rain -> ditangani pada tahap Pra-Pemrosesan.")
```

## TAHAP 3 — Pra-Pemrosesan Data *(proposal 2.3)*
**P1** penyeragaman stasiun · **P2** pembentukan & pemeriksaan tanggal · **P3** duplikat · **P4** nilai hilang (complete-case) · **P5** nilai tidak wajar · **P6** pemilihan fitur · **P7** transformasi log1p (Pers. 1) · **P8** standardisasi Z-score per stasiun (Pers. 2, parameter dari S0).


```python
# =====================================================================
# TAHAP 2.3 — PRA-PEMROSESAN DATA  (8 langkah, sesuai proposal)
# =====================================================================
log_ringkas = []   # catatan jumlah baris tiap langkah
df_pre = df.copy()

# ---------------------------------------------------------------------
# [P1] PENYERAGAMAN IDENTITAS STASIUN  (20 varian -> 5 kode: DKI1..DKI5)
# ---------------------------------------------------------------------
print("=" * 70, "\n[P1] PENYERAGAMAN IDENTITAS STASIUN\n" + "=" * 70)
df_pre["station"] = df_pre["stasiun_pengamatan"].str.extract(r"(DKI\d)")[0]
print("Varian nama sebelum :", df_pre["stasiun_pengamatan"].nunique())
print("Kategori sesudah    :", df_pre["station"].nunique(), "->", sorted(df_pre["station"].unique()))
display(df_pre["station"].value_counts().sort_index().rename("jumlah_baris").to_frame())
log_ringkas.append(("Awal", len(df_pre)))

# ---------------------------------------------------------------------
# [P2] PEMBENTUKAN & PEMERIKSAAN INFORMASI TANGGAL
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[P2] PEMBENTUKAN DAN PEMERIKSAAN TANGGAL\n" + "=" * 70)
df_pre["tanggal_dt"] = bentuk_tanggal(df_pre)
tgl_salah = df_pre[df_pre["tanggal_dt"].isna()]
print("Baris dengan tanggal tidak valid (dikeluarkan karena tidak ada koreksi valid dari sumber):")
display(tgl_salah[["periode_data", "tanggal", "stasiun_pengamatan"] + FEATURES])
df_pre = df_pre.dropna(subset=["tanggal_dt"])
print(f"Rentang tanggal valid: {df_pre['tanggal_dt'].min().date()} s.d. {df_pre['tanggal_dt'].max().date()}")
log_ringkas.append(("Setelah buang tanggal tidak valid", len(df_pre)))

# ---------------------------------------------------------------------
# [P3] PEMERIKSAAN & PENANGANAN DUPLIKAT
#      (hanya duplikat identik pada periode+tanggal+stasiun+5 variabel; nilai sama pada waktu/stasiun berbeda bukan duplikat)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[P3] DATA DUPLIKAT\n" + "=" * 70)
kunci_dup = ["periode_data", "tanggal", "station"] + FEATURES
n_dup = df_pre.duplicated(subset=kunci_dup).sum()
print("Duplikat identik:", n_dup, "-> tidak ada penghapusan" if n_dup == 0 else "-> dihapus")
# Catatan: baris yang sama persis bisa muncul sah (resolusi 30 menit tanpa kolom jam), sehingga hanya
# dihapus bila ditemukan; pada dataset ini jumlahnya 0 sehingga tidak ada baris dibuang.
df_pre = df_pre.drop_duplicates(subset=kunci_dup)
log_ringkas.append(("Setelah penanganan duplikat", len(df_pre)))

# ---------------------------------------------------------------------
# [P4] NILAI HILANG — pemeriksaan (penghapusan complete-case dijalankan setelah P5)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[P4] NILAI HILANG (COMPLETE-CASE ANALYSIS)\n" + "=" * 70)
mv_awal = df_pre[FEATURES].isna().any(axis=1).sum()
print(f"Baris dengan >=1 nilai kosong sebelum penanganan nilai tidak wajar: {mv_awal:,} "
      f"({mv_awal / len(df_pre) * 100:.2f}%) | baris lengkap: {len(df_pre) - mv_awal:,}")

# ---------------------------------------------------------------------
# [P5] PEMERIKSAAN & PENANGANAN NILAI TIDAK WAJAR
#      Nilai yang terbukti salah catat -> NaN (bukan dihapus langsung); pengamatan valid yang ekstrem dipertahankan
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[P5] NILAI TIDAK WAJAR\n" + "=" * 70)
aturan = {
    "temp_air = -39.999 (nilai sentinel/error sensor)" : ("temp_air", df_pre["temp_air"] < -30),
    "ff > 100 m/s (Feb 2025; tidak mungkin secara fisik)": ("ff", df_pre["ff"] > 100),
    "rain = 350.06 mm (identik dengan ff, salah catat)"  : ("rain", df_pre["rain"] > 300),
}
rekap_anomali = []
for nama, (kol, mask) in aturan.items():
    rekap_anomali.append({"aturan": nama, "kolom": kol, "jumlah_diubah_menjadi_NaN": int(mask.sum())})
    df_pre.loc[mask, kol] = np.nan
display(pd.DataFrame(rekap_anomali))
print("Nilai ekstrem lain yang masih masuk akal (mis. hujan lebat s.d. "
      f"{df_pre['rain'].max():.1f} mm/30 menit, radiasi s.d. {df_pre['grad'].max():.0f}) DIPERTAHANKAN.")

# Eksekusi complete-case analysis setelah anomali dijadikan NaN
n_sebelum = len(df_pre)
df_pre = df_pre.dropna(subset=FEATURES)
print(f"\nComplete-case: {n_sebelum:,} -> {len(df_pre):,} baris (dibuang {n_sebelum - len(df_pre):,})")
log_ringkas.append(("Setelah complete-case & nilai tidak wajar", len(df_pre)))

# ---------------------------------------------------------------------
# [P6] PEMILIHAN FITUR CLUSTERING
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[P6] PEMILIHAN FITUR\n" + "=" * 70)
df_clean = df_pre[["station", "tanggal_dt"] + FEATURES].reset_index(drop=True)
print("Fitur clustering :", FEATURES)
print("Informasi pendukung (bukan fitur jarak): tanggal_dt (agregasi temporal), station (pemisah lokasi)")
display(df_clean.groupby("station").size().rename("jumlah_baris_bersih").to_frame())

# ---------------------------------------------------------------------
# [P7] TRANSFORMASI DATA  X' = ln(1 + X)   (Persamaan 1)
#      Aturan: diterapkan hanya pada variabel non-negatif dengan |skewness| > 1 (berdasarkan pemeriksaan distribusi)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[P7] TRANSFORMASI LOG1P\n" + "=" * 70)
SKEW_THRESHOLD = 1.0
skew_tbl = pd.DataFrame({"skewness_sebelum": df_clean[FEATURES].skew(), "nilai_min": df_clean[FEATURES].min()})
skew_tbl["dilog"] = (skew_tbl["skewness_sebelum"].abs() > SKEW_THRESHOLD) & (skew_tbl["nilai_min"] >= 0)
LOG_COLS = skew_tbl.index[skew_tbl["dilog"]].tolist()
skew_tbl["skewness_sesudah"] = [np.log1p(df_clean[f]).skew() if f in LOG_COLS else df_clean[f].skew() for f in FEATURES]
display(skew_tbl)
print("Variabel yang ditransformasi log1p:", LOG_COLS)

def ke_ruang_fitur(frame):
    """Terapkan log1p (pada LOG_COLS) -> DataFrame fitur terubah (belum distandardisasi)."""
    out = pd.DataFrame(index=frame.index)
    for f in FEATURES:
        out[f] = np.log1p(frame[f]) if f in LOG_COLS else frame[f]
    return out

# ---------------------------------------------------------------------
# [P8] STANDARDISASI Z-SCORE per stasiun  Z = (X - μ) / σ   (Persamaan 2)
#      μ dan σ ditentukan dari baseline S0 tiap stasiun, lalu DIPAKAI ULANG untuk S1, S2, S3
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[P8] STANDARDISASI Z-SCORE (parameter dari S0 per stasiun)\n" + "=" * 70)
S0_trans = ke_ruang_fitur(df_clean)
S0_trans["station"] = df_clean["station"]
scaler_mu = S0_trans.groupby("station")[FEATURES].mean()
scaler_sd = S0_trans.groupby("station")[FEATURES].std(ddof=0)
print("Rata-rata (μ) per stasiun:");  display(scaler_mu)
print("Simpangan baku (σ) per stasiun:"); display(scaler_sd)

def standardisasi(frame):
    """Kembalikan matriks Z (DataFrame, kolom f+'_z') memakai μ,σ baseline S0 sesuai stasiun tiap baris."""
    tr = ke_ruang_fitur(frame)
    mu = scaler_mu.loc[frame["station"].values, FEATURES].values
    sd = scaler_sd.loc[frame["station"].values, FEATURES].values
    return pd.DataFrame((tr.values - mu) / sd, columns=[f + "_z" for f in FEATURES], index=frame.index)

FEATURES_Z = [f + "_z" for f in FEATURES]
S0 = pd.concat([df_clean, standardisasi(df_clean)], axis=1)      # dataset S0 (tanpa agregasi) siap pakai
print("\nVerifikasi S0: rata-rata Z per stasiun (≈0) dan simpangan baku (≈1)")
display(S0.groupby("station")[FEATURES_Z].agg(["mean", "std"]).round(3))

print("\n[Ringkasan jumlah baris tiap langkah]")
display(pd.DataFrame(log_ringkas, columns=["langkah", "jumlah_baris"]))
print("Dataset S0 siap:", S0.shape)
```

## TAHAP 4 — Agregasi Temporal *(proposal 2.4)*
Membentuk empat skenario: **S0** tanpa agregasi, **S1** harian, **S2** tiga harian, **S3** mingguan. `grad, temp_air, hum_air, ff` → *mean*; `rain` → *sum*; periode dipakai bila ≥ 75% dari acuan (48/144/336 pengamatan).

**Apa yang diagregasi?** Setiap baris data asli adalah *satu pengamatan per 30 menit* (±48 baris per stasiun per hari; dataset tidak memiliki kolom jam). Kolom `tanggal` (digabung dengan `periode_data`) menjadi dasar pembentukan *bin* waktu: **S1** = satu bin per hari, **S2** = bin 3 hari berurutan, **S3** = bin 7 hari berurutan, dihitung terpisah untuk tiap stasiun. Semua baris dalam satu bin digabung menjadi **satu baris baru** (5 variabel meteorologi), sehingga jumlah observasi mengecil dari ±45.000 (S0) menjadi ±900 (S1), ±300 (S2), ±130 (S3) per stasiun. Hasilnya **4 data frame** yang berbeda.


```python
# =====================================================================
# TAHAP 2.4 — AGREGASI TEMPORAL  (S0, S1, S2, S3)
# =====================================================================
# Aturan agregasi (sesuai proposal):
#   - grad, temp_air, hum_air, ff -> MEAN ; rain -> SUM
#   - periode berurutan, tidak saling tumpang tindih (blok 1/3/7 hari dihitung dari tanggal awal data)
#   - periode dipakai bila jumlah pengamatan valid >= 75% dari acuan (48 baris/hari)
#   - transformasi log1p & standardisasi memakai parameter S0 per stasiun (dari tahap Pra-Pemrosesan)
OBS_PER_HARI = 48
MIN_RATIO    = 0.75
SKENARIO = {            # kode: (nama, lebar periode dalam hari)
    "S0": ("Tanpa agregasi", None),
    "S1": ("Harian (1 hari)", 1),
    "S2": ("Tiga harian (3 hari)", 3),
    "S3": ("Mingguan (7 hari)", 7),
}
TGL_MULAI = df_clean["tanggal_dt"].min()
agg_fungsi = {f: ("sum" if f == "rain" else "mean") for f in FEATURES}

def agregasi_temporal(frame, lebar_hari):
    f = frame.copy()
    f["periode"] = (f["tanggal_dt"] - TGL_MULAI).dt.days // lebar_hari
    g = f.groupby(["station", "periode"], sort=True)
    hasil = g.agg(**{c: (c, fn) for c, fn in agg_fungsi.items()},
                  n_valid=("grad", "size"), tanggal_mulai=("tanggal_dt", "min")).reset_index()
    acuan = OBS_PER_HARI * lebar_hari
    hasil["acuan"] = acuan
    lolos = hasil["n_valid"] >= MIN_RATIO * acuan
    return hasil, hasil[lolos].reset_index(drop=True)

scenarios = {"S0": S0.copy()}
rekap = [{"skenario": "S0", "stasiun": s, "periode_terbentuk": n, "periode_lolos_75%": n, "periode_dibuang": 0, "acuan_obs": 1, "min_valid": 1}
         for s, n in S0.groupby("station").size().items()]
for kode, (nama, lebar) in SKENARIO.items():
    if lebar is None:
        continue
    semua, lolos = agregasi_temporal(df_clean, lebar)
    for s in sorted(semua["station"].unique()):
        a, b = (semua["station"] == s).sum(), (lolos["station"] == s).sum()
        rekap.append({"skenario": kode, "stasiun": s, "periode_terbentuk": a, "periode_lolos_75%": b, "periode_dibuang": a - b,
                      "acuan_obs": OBS_PER_HARI * lebar, "min_valid": int(np.ceil(MIN_RATIO * OBS_PER_HARI * lebar))})
    scenarios[kode] = pd.concat([lolos, standardisasi(lolos)], axis=1)   # μ,σ S0 dipakai ulang

rekap = pd.DataFrame(rekap)
print("=" * 70, "\nREKAP KELENGKAPAN PERIODE (aturan 75%)\n" + "=" * 70)
display(rekap)

print("\nJumlah observasi tiap skenario (seluruh stasiun):")
ringkas_n = pd.DataFrame({k: v.groupby("station").size() for k, v in scenarios.items()})
ringkas_n.loc["TOTAL"] = ringkas_n.sum()
display(ringkas_n)

print("\nStatistik deskriptif (satuan asli) per skenario — dampak agregasi terhadap sebaran nilai:")
display(pd.concat({k: v[FEATURES].describe().T[["mean", "std", "min", "max"]] for k, v in scenarios.items()}))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
ringkas_n.drop("TOTAL").T.plot(kind="bar", ax=axes[0], rot=0, colormap="tab10")
axes[0].set_yscale("log"); axes[0].set_title("Jumlah observasi per skenario & stasiun (skala log)"); axes[0].set_ylabel("n observasi")
for k in ["S0", "S1", "S2", "S3"]:
    axes[1].hist(scenarios[k]["temp_air_z"], bins=60, alpha=0.5, density=True, label=k)
axes[1].set_title("Sebaran temp_air (Z-score) per skenario"); axes[1].legend()
plt.tight_layout(); plt.show()

print("\nKeempat dataset siap pada dict `scenarios`:", {k: v.shape for k, v in scenarios.items()})

# ---------------------------------------------------------------------
# [4.X] TABEL 3 — CONTOH HASIL AGREGASI PADA SKENARIO S1, S2, DAN S3
#       (periode pertama yang lolos aturan 75% pada stasiun contoh; satuan asli,
#        rain = jumlah per periode, empat variabel lain = rata-rata)
# ---------------------------------------------------------------------
STASIUN_CONTOH = sorted(scenarios["S1"]["station"].unique())[0]
contoh_baris = []
for kode, (nama, lebar) in SKENARIO.items():
    if lebar is None:
        continue
    r = scenarios[kode][scenarios[kode]["station"] == STASIUN_CONTOH].iloc[0]
    awal  = TGL_MULAI + pd.Timedelta(days=int(r["periode"]) * lebar)
    akhir = awal + pd.Timedelta(days=lebar - 1)
    contoh_baris.append({
        "Skenario": kode,
        "Stasiun": r["station"],
        "Periode": f"{awal:%d-%m-%Y}" if lebar == 1 else f"{awal:%d-%m-%Y} s.d. {akhir:%d-%m-%Y}",
        "Pengamatan valid": f"{int(r['n_valid'])} (min. {int(np.ceil(MIN_RATIO * r['acuan']))})",
        "grad": round(r["grad"], 2), "temp_air": round(r["temp_air"], 2),
        "hum_air": round(r["hum_air"], 2), "ff": round(r["ff"], 2), "rain": round(r["rain"], 2),
    })
tabel3 = pd.DataFrame(contoh_baris)
print("\n[Tabel 3] Contoh hasil agregasi pada skenario S1, S2, dan S3")
display(tabel3)
tabel3.to_csv(os.path.join(OUT_DIR, "tabel3_contoh_hasil_agregasi.csv"), index=False, encoding="utf-8-sig")
```

## TAHAP 5 — Penerapan Algoritma DBSCAN *(proposal 2.5)*
**D1** parameter dikunci sebagai variabel kontrol (ε = 0,5; MinPts = 5, default scikit-learn; k-distance graph hanya sebagai referensi) · **D2** DBSCAN dengan parameter yang sama pada S0–S3 untuk tiap stasiun; label `-1` = noise; waktu komputasi diukur.


```python
# =====================================================================
# TAHAP 2.5 — PENERAPAN ALGORITMA DBSCAN
# =====================================================================
STATIONS = sorted(S0["station"].unique())

# ---------------------------------------------------------------------
# [D1] PARAMETER DBSCAN — VARIABEL KONTROL (dikunci, tidak diubah antarskenario)
#      ε = 0.5 dan MinPts = 5 (nilai default scikit-learn; ditetapkan di Tahap 1)
#      Faktor yang diubah HANYA tingkat agregasi temporal (S0, S1, S2, S3).
# ---------------------------------------------------------------------
print("=" * 70, "\n[D1] PARAMETER DBSCAN (VARIABEL KONTROL)\n" + "=" * 70)
print(f"ε = {EPS} | MinPts = {MIN_PTS}  -> sama untuk seluruh skenario dan seluruh stasiun")

# k-distance graph pada S0: hanya sebagai GAMBARAN KARAKTERISTIK DISTRIBUSI data (tidak dipakai mengubah parameter)
kdist = {}
for s in STATIONS:
    Z = S0.loc[S0["station"] == s, FEATURES_Z].values
    d, _ = NearestNeighbors(n_neighbors=MIN_PTS, n_jobs=-1).fit(Z).kneighbors(Z)
    kdist[s] = np.sort(d[:, -1])
fig, ax = plt.subplots(figsize=(10, 5))
for s in STATIONS:
    ax.plot(kdist[s], lw=1.2, label=f"{s} ({(kdist[s] <= EPS).mean() * 100:.1f}% titik ≤ ε)")
ax.axhline(EPS, color="k", ls="--", lw=1, label=f"ε tetap = {EPS}")
ax.set_ylim(0, np.percentile(np.concatenate(list(kdist.values())), 99.5))
ax.set_title(f"k-distance graph pada S0 (k = MinPts = {MIN_PTS}) — referensi, bukan penentu parameter")
ax.set_xlabel("titik data (terurut)"); ax.set_ylabel(f"jarak ke tetangga ke-{MIN_PTS}"); ax.legend()
plt.tight_layout(); plt.show()

# ---------------------------------------------------------------------
# [D2] MENJALANKAN DBSCAN pada 4 skenario x 5 stasiun (parameter sama), ukur waktu komputasi
#      Label -1 = noise (tidak memenuhi kepadatan minimum; BUKAN otomatis cuaca ekstrem/data salah)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[D2] PENERAPAN DBSCAN PADA SELURUH SKENARIO\n" + "=" * 70)
hasil_dbscan = []
for kode, frame in scenarios.items():
    frame["label"] = -2                               # penanda sementara
    for s in STATIONS:
        idx = frame.index[frame["station"] == s]
        Z = frame.loc[idx, FEATURES_Z].values
        t0 = time.perf_counter()
        lab = DBSCAN(eps=EPS, min_samples=MIN_PTS, n_jobs=-1).fit_predict(Z)
        waktu = time.perf_counter() - t0
        frame.loc[idx, "label"] = lab
        hasil_dbscan.append({"skenario": kode, "stasiun": s, "n_total": len(idx),
                             "n_cluster": len(set(lab) - {-1}), "n_noise": int((lab == -1).sum()),
                             "noise_pct": (lab == -1).sum() / len(idx) * 100,       # Persamaan (6)
                             "waktu_detik": waktu})
    print(f"  {kode} selesai.")
hasil_dbscan = pd.DataFrame(hasil_dbscan)
assert all((f["label"] != -2).all() for f in scenarios.values())

print("\nLabel cluster, jumlah cluster, jumlah noise, dan waktu komputasi tiap skenario & stasiun:")
display(hasil_dbscan)
print("Ringkasan per skenario (jumlah cluster & waktu: rata-rata antarstasiun; noise: gabungan seluruh stasiun):")
display(hasil_dbscan.groupby("skenario").agg(n_total=("n_total", "sum"), n_cluster_rata2=("n_cluster", "mean"),
                                              n_noise=("n_noise", "sum"), waktu_total_detik=("waktu_detik", "sum")))
```

## TAHAP 6 — Evaluasi Hasil Clustering *(proposal 2.6)*
**E1** Silhouette Score (Pers. 4) & Davies-Bouldin Index (Pers. 5), tanpa noise dan bila ≥ 2 cluster · **E2** jumlah cluster, % noise (Pers. 6), waktu komputasi · **E3** visualisasi.


```python
# =====================================================================
# TAHAP 2.6 — EVALUASI HASIL CLUSTERING
# =====================================================================
# Metrik: Silhouette Score (Pers. 4), Davies-Bouldin Index (Pers. 5), jumlah cluster, % noise (Pers. 6), waktu komputasi.
# Silhouette & DBI dihitung HANYA pada data non-noise dan bila terbentuk >= 2 cluster valid (selain itu = NaN).
print("=" * 70, "\n[E1] SILHOUETTE SCORE & DAVIES-BOULDIN INDEX\n" + "=" * 70)

def evaluasi_cluster(Z, lab):
    ok = lab != -1
    n_cl = len(set(lab[ok]))
    if n_cl < 2 or ok.sum() <= n_cl:
        return np.nan, np.nan
    Zk, Lk = Z[ok], lab[ok]
    ukuran = min(SIL_SAMPLE, len(Zk)) if SIL_SAMPLE else None
    sil = silhouette_score(Zk, Lk, sample_size=ukuran, random_state=RANDOM_STATE)
    dbi = davies_bouldin_score(Zk, Lk)
    return sil, dbi

baris = []
for kode, frame in scenarios.items():
    for s in STATIONS:
        sub = frame[frame["station"] == s]
        sil, dbi = evaluasi_cluster(sub[FEATURES_Z].values, sub["label"].values)
        baris.append({"skenario": kode, "stasiun": s, "silhouette": sil, "dbi": dbi})
metrik_kualitas = pd.DataFrame(baris)

# ---------------------------------------------------------------------
# [E2] GABUNGAN SELURUH METRIK PER SKENARIO x STASIUN
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[E2] TABEL EVALUASI LENGKAP\n" + "=" * 70)
eval_df = hasil_dbscan.merge(metrik_kualitas, on=["skenario", "stasiun"])
eval_df = eval_df[["skenario", "stasiun", "n_total", "n_cluster", "n_noise", "noise_pct", "silhouette", "dbi", "waktu_detik"]]
display(eval_df)

print("Ringkasan per skenario (rata-rata antarstasiun; NaN diabaikan):")
eval_skenario = eval_df.groupby("skenario").agg(
    n_cluster=("n_cluster", "mean"), noise_pct=("noise_pct", "mean"),
    silhouette=("silhouette", "mean"), dbi=("dbi", "mean"), waktu_detik=("waktu_detik", "mean"),
    stasiun_valid=("silhouette", "count"))
display(eval_skenario)
print("Catatan: 'stasiun_valid' = jumlah stasiun yang membentuk >= 2 cluster sehingga Silhouette/DBI dapat dihitung.")
print(f"Silhouette dihitung dengan sampel acak maks. {SIL_SAMPLE} titik non-noise per stasiun (SIL_SAMPLE={SIL_SAMPLE}, seed={RANDOM_STATE}).")

# ---------------------------------------------------------------------
# [E3] VISUALISASI METRIK EVALUASI
# ---------------------------------------------------------------------
metrik_plot = [("silhouette", "Silhouette Score (↑ lebih baik)"), ("dbi", "Davies-Bouldin Index (↓ lebih baik)"),
               ("n_cluster", "Jumlah cluster"), ("noise_pct", "Persentase noise (%)"), ("waktu_detik", "Waktu komputasi (detik)")]
fig, axes = plt.subplots(2, 3, figsize=(16, 8))
for ax, (m, judul) in zip(axes.ravel(), metrik_plot):
    sns.barplot(data=eval_df, x="skenario", y=m, hue="stasiun", palette="colorblind", ax=ax)
    ax.set_title(judul); ax.set_xlabel("")
    if ax is not axes.ravel()[0]:
        ax.get_legend().remove()
axes.ravel()[-1].axis("off")
plt.tight_layout(); plt.show()

eval_df.to_csv(os.path.join(OUT_DIR, "evaluasi_per_stasiun.csv"), index=False)
eval_skenario.to_csv(os.path.join(OUT_DIR, "evaluasi_per_skenario.csv"))
print(f"Tabel evaluasi disimpan di folder '{OUT_DIR}/'.")
```

## TAHAP 7 — Analisis dan Perbandingan Hasil Clustering *(proposal 2.7)*
**F1** perbandingan metrik antarskenario · **F2** delta terhadap baseline ΔM = M<sub>i</sub> − M<sub>0</sub> (Pers. 7) · **F3** skenario terbaik · **F4** karakteristik cluster (rata-rata & distribusi 5 variabel) pada skenario terbaik · **F5** ringkasan.


```python
# =====================================================================
# TAHAP 2.7 — ANALISIS DAN PERBANDINGAN HASIL CLUSTERING
# =====================================================================
METRIK = ["silhouette", "dbi", "noise_pct", "n_cluster", "waktu_detik"]

# ---------------------------------------------------------------------
# [F1] PERBANDINGAN ANTAR SKENARIO
# ---------------------------------------------------------------------
print("=" * 70, "\n[F1] PERBANDINGAN METRIK ANTAR SKENARIO (S0 s.d. S3)\n" + "=" * 70)
display(eval_skenario[["silhouette", "dbi", "n_cluster", "noise_pct", "waktu_detik", "stasiun_valid"]])
print("Catatan: Silhouette/DBI dirata-ratakan hanya pada stasiun yang valid (lihat 'stasiun_valid'); baca bersama % noise.")

# ---------------------------------------------------------------------
# [F2] DELTA TERHADAP BASELINE   ΔM_i = M_i − M_0   (Persamaan 7)
#      Silhouette: Δ>0 membaik | DBI: Δ<0 membaik | noise/cluster/waktu: dibaca sebagai perubahan apa adanya
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[F2] DELTA TERHADAP BASELINE S0 (ΔM = M_skenario − M_S0)\n" + "=" * 70)
base = eval_df[eval_df["skenario"] == "S0"].set_index("stasiun")[METRIK]
delta = eval_df[eval_df["skenario"] != "S0"].copy()
for m in METRIK:
    delta["Δ" + m] = delta.apply(lambda r: r[m] - base.loc[r["stasiun"], m], axis=1)
delta_stasiun = delta[["skenario", "stasiun"] + ["Δ" + m for m in METRIK]]
print("Delta per stasiun:")
display(delta_stasiun)

delta_skenario = delta.groupby("skenario")[["Δ" + m for m in METRIK]].mean()
print("Delta rata-rata antarstasiun (NaN diabaikan):")
display(delta_skenario)

def arah(m, v):
    if pd.isna(v): return "tidak dapat dihitung"
    if m == "silhouette": return "membaik" if v > 0 else "menurun"
    if m == "dbi":        return "membaik" if v < 0 else "memburuk"
    return "naik" if v > 0 else "turun"
tafsir = pd.DataFrame({m: [arah(m, delta_skenario.loc[k, "Δ" + m]) for k in delta_skenario.index] for m in METRIK},
                      index=delta_skenario.index)
print("Interpretasi arah perubahan (rata-rata antarstasiun):")
display(tafsir)

# Skenario yang paling mendekati baseline = |Δ| terkecil (metrik dinormalkan dengan nilai S0 rata-rata)
ref = eval_skenario.loc["S0", METRIK].replace(0, np.nan)
jarak = (delta_skenario[["Δ" + m for m in ["silhouette", "dbi", "noise_pct", "n_cluster"]]].abs()
         .div(ref[["silhouette", "dbi", "noise_pct", "n_cluster"]].values)).mean(axis=1).rename("selisih_relatif_rata2")
print("Kedekatan terhadap baseline (selisih relatif rata-rata; makin kecil = makin mirip S0):")
display(jarak.sort_values().to_frame())
print(f">> Skenario paling mendekati baseline S0: {jarak.idxmin()}")

fig, axes = plt.subplots(1, 5, figsize=(18, 3.8))
for ax, m in zip(axes, METRIK):
    vals = delta_skenario["Δ" + m]
    ax.bar(vals.index, vals.values, color=sns.color_palette("colorblind", 3))
    ax.axhline(0, color="k", lw=0.8); ax.set_title("Δ " + m)
plt.suptitle("Perubahan rata-rata metrik terhadap baseline S0", y=1.03); plt.tight_layout(); plt.show()

# ---------------------------------------------------------------------
# [F3] PEMILIHAN SKENARIO TERBAIK: rata-rata peringkat (Silhouette tertinggi, DBI terendah); bila imbang -> noise lebih kecil
#      Syarat kelayakan: metrik valid pada >= MIN_STASIUN_VALID stasiun (agar rata-rata tidak bias oleh 1 stasiun saja)
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[F3] SKENARIO DENGAN KUALITAS TERBAIK\n" + "=" * 70)
MIN_STASIUN_VALID = 3     # skenario hanya layak dibandingkan bila Silhouette/DBI dapat dihitung pada >= 3 dari 5 stasiun
kandidat = eval_skenario.dropna(subset=["silhouette", "dbi"]).copy()
tidak_layak = kandidat[kandidat["stasiun_valid"] < MIN_STASIUN_VALID]
if len(tidak_layak):
    print(f"Skenario {list(tidak_layak.index)} TIDAK diikutkan dalam pemilihan: metrik hanya valid pada "
          f"{tidak_layak['stasiun_valid'].astype(int).tolist()} stasiun (< {MIN_STASIUN_VALID}); mayoritas data menjadi noise "
          f"(noise {tidak_layak['noise_pct'].round(1).tolist()}%) dengan ε/MinPts yang dikunci sama untuk semua skenario.")
kandidat = kandidat[kandidat["stasiun_valid"] >= MIN_STASIUN_VALID]
if kandidat.empty:
    raise RuntimeError("Tidak ada skenario dengan metrik valid pada cukup stasiun; ubah EPS/MIN_PTS di Tahap 1.")
kandidat["rank_sil"] = kandidat["silhouette"].rank(ascending=False)
kandidat["rank_dbi"] = kandidat["dbi"].rank(ascending=True)
kandidat["skor_rank"] = (kandidat["rank_sil"] + kandidat["rank_dbi"]) / 2
kandidat = kandidat.sort_values(["skor_rank", "noise_pct"])
display(kandidat[["silhouette", "dbi", "noise_pct", "rank_sil", "rank_dbi", "skor_rank"]])
BEST = kandidat.index[0]
print(f">> Skenario terbaik: {BEST} — {SKENARIO[BEST][0]}")
print("   (Noise yang lebih kecil tidak otomatis berarti lebih baik; dibaca bersama Silhouette & DBI.)")

# ---------------------------------------------------------------------
# [F4] KARAKTERISTIK CLUSTER PADA SKENARIO TERBAIK
#      Rata-rata & distribusi 5 variabel (satuan asli skenario terbaik) pada tiap cluster; noise ditampilkan terpisah
# ---------------------------------------------------------------------
print("\n" + "=" * 70, f"\n[F4] KARAKTERISTIK CLUSTER — SKENARIO {BEST}\n" + "=" * 70)
best_df = scenarios[BEST].copy()
best_df["cluster"] = best_df["label"].map(lambda v: "noise" if v == -1 else f"C{v}")
profil_semua = []
for s in STATIONS:
    sub = best_df[best_df["station"] == s]
    prof = sub.groupby("cluster").agg(n=("cluster", "size"), **{f: (f, "mean") for f in FEATURES})
    prof["persen"] = prof["n"] / prof["n"].sum() * 100
    prof = prof.sort_values("n", ascending=False)
    prof.insert(0, "stasiun", s)
    profil_semua.append(prof.reset_index())
    print(f"\n--- Stasiun {s}: {sub['label'].nunique() - (1 if (sub['label'] == -1).any() else 0)} cluster, "
          f"noise {(sub['label'] == -1).mean() * 100:.2f}% ---")
    display(prof.drop(columns="stasiun"))
    # heatmap profil (Z-score rata-rata cluster pada ruang fitur) — hanya cluster berukuran >= 1% atau noise
    prof_z = sub.groupby("cluster")[FEATURES_Z].mean().loc[prof.index]
    prof_z.columns = FEATURES
    top = prof.index[:10]                                          # maks 10 kelompok terbesar agar terbaca
    # --- boxplot distribusi tiap variabel per cluster
    fig, axes = plt.subplots(1, 6, figsize=(20, 3.8), gridspec_kw={"width_ratios": [1.3, 1, 1, 1, 1, 1]})
    sns.heatmap(prof_z.loc[top], annot=True, fmt=".2f", cmap="RdBu_r", center=0, ax=axes[0], cbar=False)
    axes[0].set_title(f"{s}: rata-rata Z per cluster"); axes[0].set_ylabel("")
    for ax, f in zip(axes[1:], FEATURES):
        sns.boxplot(data=sub[sub["cluster"].isin(top)], x="cluster", y=f, order=list(top), showfliers=False,
                    color="#8FB3D9", ax=ax)
        ax.set_title(f); ax.set_xlabel(""); ax.tick_params(axis="x", rotation=60)
    plt.tight_layout(); plt.show()

profil_semua = pd.concat(profil_semua, ignore_index=True)
profil_semua.to_csv(os.path.join(OUT_DIR, f"profil_cluster_{BEST}.csv"), index=False)
delta_stasiun.to_csv(os.path.join(OUT_DIR, "delta_vs_baseline.csv"), index=False)

# ---------------------------------------------------------------------
# [F5] RINGKASAN HASIL EKSPERIMEN
# ---------------------------------------------------------------------
print("\n" + "=" * 70, "\n[F5] RINGKASAN HASIL\n" + "=" * 70)
print(f"Parameter DBSCAN (kontrol, tetap) : ε = {EPS}, MinPts = {MIN_PTS}")
print(f"Skenario kualitas terbaik         : {BEST} ({SKENARIO[BEST][0]})")
for k in eval_skenario.index:
    r = eval_skenario.loc[k]
    print(f"  {k}: cluster≈{r['n_cluster']:.1f} | noise={r['noise_pct']:.2f}% | Silhouette={r['silhouette']:.3f} | "
          f"DBI={r['dbi']:.3f} | waktu={r['waktu_detik']:.2f} dtk | stasiun valid={int(r['stasiun_valid'])}/5")
print(f"\nSemua tabel hasil tersimpan di folder '{OUT_DIR}/'.")
```
